```python
lista =[10, 13, 14, 18, 22]
```

```python
def media(lista):
  return sum(lista) / len(lista)
```

```python
media(lista)
```

(sumatoria de todos los datos - media ) ** 2/ n

```python
def desviacion_estandar(lista):
  for i in lista:
    return ((lista)[i] - media(lista)) ** 2/len(lista)
```
